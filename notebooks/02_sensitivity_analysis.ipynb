# 02 — Monte Carlo scenario sensitivity

The analysis varies a **hypothetical duration multiplier** and random seeds. The output describes a conditional model response, **not** empirical forecasting uncertainty. Run the README sensitivity command first.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

path = Path("output/sensitivity/barrido_v32.csv")
if not path.exists():
    raise FileNotFoundError("Run the sensitivity command in README.md first")
df = pd.read_csv(path)
assert {"scale_hypothetical", "seed", "autonomy_by_maximum"}.issubset(df.columns)
df.head()
```

## Summary across seeds

```python
summary = df.groupby("scale_hypothetical")["autonomy_by_maximum"].agg(["count", "min", "median", "max"])
summary.round(3)
```

```python
fig, ax = plt.subplots(figsize=(9,4.5))
x=summary.index.to_numpy(dtype=float)
ax.fill_between(x, summary["min"].to_numpy(), summary["max"].to_numpy(), alpha=0.2, label="Min–max across selected seeds")
ax.plot(x, summary["median"].to_numpy(), marker="o", label="Median simulated proportion")
ax.set(xlabel="Assumed operational duration multiplier", ylabel="Synthetic autonomy by assumed maximum", ylim=(0,1), title="Sensitivity to hypothetical operational durations")
ax.legend()
plt.show()
```

## Category-specific results

Compare Category I/II/III against different **assumed** deadline thresholds. The categories are produced by artificial scoring rules.

```python
columns=[c for c in df.columns if c.startswith("category_")]
df.groupby("scale_hypothetical")[columns].median().round(3)
```

## Critical interpretation

The shaded range is **not a 95% confidence interval**. These are selected seed outcomes conditional on the chosen distributions; neither distributional uncertainty nor historical sampling bias is resolved by adding more seeds. A high simulated success rate can be caused by permissive assumed deadlines.
