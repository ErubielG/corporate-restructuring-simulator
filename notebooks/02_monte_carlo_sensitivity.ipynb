# 02 · Monte Carlo sensitivity analysis

This exercise varies an **assumed operational-duration multiplier** across repeated
pseudo-random seeds. The band is the *min/max across chosen seeds*, **not** a confidence
interval for a population of real companies.

Run `python scripts/run_sensitivity.py --out output/sensitivity/sensitivity.csv` first.

```
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().resolve()
if not (ROOT / "config_demo_synthetic.json").exists():
    ROOT = ROOT.parent
path = ROOT / "output/sensitivity/sensitivity.csv"
assert path.exists(), "Generate sensitivity.csv using scripts/run_sensitivity.py first"
results = pd.read_csv(path)
results.head()
```

```
grouped = results.groupby("operational_scale_hypothetical")["autonomy_by_maximum_simulated"]
summary = grouped.agg(["min", "median", "max", "count"])
summary
```

```
ax = summary[["min", "median", "max"]].plot(marker="o", figsize=(8, 4))
ax.set(title="Sensitivity to hypothetical operational duration", xlabel="Operational scale factor", ylabel="Simulated share within artificial maximum", ylim=(0, 1.02))
plt.show()
```

## Questions to investigate

- Does increasing the assumed operational scale change all categories equally?
- Is there a noticeable interaction between the scale and the artificial deadlines?
- Does variation across a handful of seeds quantify uncertainty about the **real-world**
  input distributions? **No.** Changing stochastic seeds explores run-to-run variability
  *conditional on the simulation model*; the calibration uncertainty remains unresolved.
